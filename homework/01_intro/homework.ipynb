## Q6. Median value of horsepower

- Find the median value of the horsepower column in the dataset.
- Next, calculate the most frequent value of the same horsepower column.
- Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
- Now, calculate the median value of horsepower once again.

Has it changed?

```python
median_hp = df['horsepower'].median()
median_hp
```

```python
mode_hp = df['horsepower'].mode()[0]
mode_hp
```

```python
df.loc[:, 'horsepower'] = df['horsepower'].fillna(mode_hp)
```

```python
df['horsepower'].isna().sum()
```

```python
median_hp_imputed = df['horsepower'].median()
median_hp_imputed
```

```python
if median_hp_imputed > median_hp:
    has_changed = "increased"
elif median_hp_imputed < median_hp:
    has_changed = "decreased"
elif median_hp_imputed == median_hp:
    has_changed = "no change"

print(f"Change after fillna: {has_changed}")
```

## Q7: Sum of weights

1. Select all the cars from Asia
1. Select only columns vehicle_weight and model_year
1. Select the first 7 values
1. Get the underlying NumPy array. Let's call it X.
1. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
1. Invert XTX.
1. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
1. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
1. What's the sum of all the elements of the result?

```python
df_asia = df[df['origin']=='Asia']
```

```python
df_asia.info()
```

```python
df_asia_weight_year = df_asia[['vehicle_weight', 'model_year']]
```

```python
df_top_7 = df_asia_weight_year.head(7)
```

```python
X = df_top_7.to_numpy()
```

```python
XTX = np.dot(X.T, X)
```

```python
XTX
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = np.dot(
    np.dot(np.linalg.inv(XTX), X.T), y
)
```

```python
w.sum()
```
